## Combinatorial Numbers

Implement a function that calculates binomial coefficients (also known as elements of Pascal's Triangle) up to a depth of $N$. Print them in a "triangle format".

The binomial coefficient is defined mathematically as:

$\frac{n!}{k!(n−k)!}$

Where $n!$ is the factorial of $n$.

For Pascal’s Triangle, the structure starts with 1 at the top, and every row $n$ contains the binomial coefficients for that row, iterating over $k$ from $0$ to $n$ for computing the elements of the row:

```
      1       <- Row 0: 1 element (n=0, k=0)
     1 1      <- Row 1: 2 elements (n=1, k=0,1)
    1 2 1     <- Row 2: 3 elements (n=2, k=0,1,2)
   1 3 3 1    <- Row 3: 4 elements (n=3, k=0,1,2,3)
  1 4 6 4 1   <- Row 4: 5 elements (n=4, k=0,1,2,3,4)
```

Write a function called `pascal(N)` for displaying Pascal's triangle up to a given number depth of $N$, which specifies the number of rows of the triangle.

You can use `factorial()` function from `math` module to compute the factorial of a number.

````python
from math import factorial

def pascal(N):
    for n in range(N):
        print(" " * (N - n), end="")

        for k in range(n + 1):
            coef = factorial(n) // (factorial(k) * factorial(n - k))
            print(coef, end=" ")
        ks = ",".join(str(value) for value in range(n +1))
        print(f" <- Row {n}: {n+1} elements (n={n}, k={ks})")
````

````python
pascal(5)
````

## Robot Movement

- A robot moves on a plane starting from point (0,0).
- The robot can move UP, DOWN, LEFT, and RIGHT.
- The robot's movement trail is given as input string like: `UP 5 DOWN 3 LEFT 3 RIGHT 2`, where the numbers after the direction are the number of steps in that direction.

Write a program to calculate the Euclidean distance from the current position after a sequence of movements to the original point. The Euclidean distance between two points (x1, y1) and (x2, y2) is defined as:

    Euclidean distance = SQRT((x1 - x2)^2 + (y1 - y2)^2),

where SQRT is the square root (√).

Example: 
- Input: UP 5 DOWN 3 LEFT 3 RIGHT 2
- Output: 2.24 (rounded to 2 decimals)

````python
import math

def calculate_robot_distance(movement_string: str) -> float:
    x, y = 0, 0

    parts = movement_string.split()

    for i in range(0, len(parts), 2):
        direction = parts[i].upper()
        steps = int(parts[i+1])

        if direction == "UP":
            y += steps
        elif direction == "DOWN":
            y -= steps
        elif direction == "LEFT":
            x -= steps
        elif direction == "RIGHT":
            x += steps

    x1, y1 = int(x), int(y)
    x2, y2 = int(0), int(0)

    distance = math.sqrt((x1 - x2)**2 + (y1 - y2)**2)

    return round(distance, 2)


if __name__ == "__main__":
    input_trail = "UP 5 DOWN 3 LEFT 3 RIGHT 2"
    result = calculate_robot_distance(input_trail)

    print(f"Trayectoria: {input_trail}")
    print(f"Distancia final al origen: {result}")
````

## Color by Proximity — Nearest Neighbors

Write a function:

`predict_color(points, new_point, k)`

that **predicts** the color of a new point in 2D space based on its **k** nearest neighbors.

Each point is represented as a tuple `(x, y, color)`, where:

- `x` and `y` are coordinates,
- `color` is a string (e.g. "red", "blue", "green").

The function should:

1. Compute the **Euclidean distance** between the new point and each existing point.
2. Select the **k points with the smallest distances**.
3. Predict (return) the color as the **majority color** among those k neighbors.
    - If there’s a tie between two or more colors, you can return any of them.


This is a simplified version of a ML algorithm called k-Nearest Neighbors (kNN) that we will see in class.

````python
# Example data
points = [
    (1.5, 1.8, "red"),
    (2.0, 2.1, "red"),
    (2.2, 1.9, "red"),
    (1.8, 2.3, "red"),
    (2.1, 2.4, "red"),
    (2.7, 1.8, "green"),
    (3.0, 2.2, "green"),
    (3.2, 1.9, "green"),
    (2.8, 2.4, "green"),
    (3.1, 2.5, "green"),
    (3.3, 2.8, "blue"),
    (3.5, 3.1, "blue"),
    (3.7, 2.9, "blue"),
    (3.4, 3.3, "blue"),
    (3.6, 3.2, "blue"),
]
````

````python
# These are examples only, here we can have any other value
new_point = (3, 2.8)
k = 5
````

````python
# Write your function here
import math
from collections import Counter


def predict_color(points, new_point, k):
    # 1. Calcular la distancia Euclídea entre el 'new_point' y cada punto existente
    distances = []
    x_new, y_new = new_point

    for x, y, color in points:
        # Fórmula de la distancia euclídea en 2D: sqrt((x2 - x1)^2 + (y2 - y1)^2)
        dist = math.sqrt((x - x_new) ** 2 + (y - y_new) ** 2)
        distances.append((dist, color))

    # 2. Ordenar las distancias de menor a mayor
    # Como guardamos tuplas (distancia, color), ordenará automáticamente por la distancia
    distances.sort(key=lambda item: item[0])

    # 3. Seleccionar los 'k' vecinos más cercanos (los primeros k elementos)
    k_nearest = distances[:k]

    # Extraer solo los colores de esos k vecinos
    k_colors = [color for dist, color in k_nearest]

    # 4. Predecir el color mayoritario usando Counter (cuenta cuántas veces aparece cada color)
    color_counts = Counter(k_colors)

    # .most_common(1)[0][0] nos da el elemento con la frecuencia más alta (el ganador)
    majority_color = color_counts.most_common(1)[0][0]

    return majority_color


# --- Pruebas de funcionamiento con tus datos de ejemplo ---
points = [
    (1.5, 1.8, "red"),
    (2.0, 2.1, "red"),
    (2.2, 1.9, "red"),
    (1.8, 2.3, "red"),
    (2.1, 2.4, "red"),
    (2.7, 1.8, "green"),
    (3.0, 2.2, "green"),
    (3.2, 1.9, "green"),
    (2.8, 2.4, "green"),
    (3.1, 2.5, "green"),
    (3.3, 2.8, "blue"),
    (3.5, 3.1, "blue"),
    (3.7, 2.9, "blue"),
    (3.4, 3.3, "blue"),
    (3.6, 3.2, "blue"),
]

# Probemos a predecir el color de un punto nuevo rodeado de puntos rojos
un_punto_nuevo = (1.9, 2.0)
resultado = predict_color(points, un_punto_nuevo, k=3)
print(f"El color predicho para el punto {un_punto_nuevo} es: {resultado}")
````

````python
# This code assumes you've implemented the `predict_color` function that returns a color prediction

import matplotlib.pyplot as plt

predicted_color = predict_color(points, new_point, k)
print("Predicted color:", predicted_color)

plt.figure(figsize=(6,6))
for x, y, color in points:
    plt.scatter(x, y, c=color, s=100, label=color if color not in plt.gca().get_legend_handles_labels()[1] else "")

plt.scatter(new_point[0], new_point[1], c=predicted_color, s=250, edgecolors='black', marker='X', label="Prediction")
plt.legend()
plt.title(f"k = {k}, predicted color = {predicted_color}")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
````

## kNN Bonus

Extend the previous **Color by proximity** exercise by implementing the following extras:

1. **Distance function**: Add another parameter to the `predict_color()` function called `dist_fn`, which represents the distance function used to calculate how close two points are.

    Previously, you "hardcoded" the Euclidean distance, but now your function should use whatever distance function is passed as a parameter — so a programmer can choose the distance metric.

    For example, try computing distances using the **Manhattan distance**:
        `Manhattan distance = |x1 - x2| + |y1 - y2|`

2. **Weighted vote**: Instead of a simple majority vote for predicting the color, give **closer neighbors higher influence**. The idea is that nearer points should “count more” toward the final prediction.

    You can assign each neighbor a weight inversely proportional to its distance, such as: `weight = 1 / distance`. To handle the case where the distance is 0 (the new point exactly matches an existing one), you can return the color of that point immediately.

    Your goal is to sum the weights of the neighbors for each color and predict the color with the largest total weight.

````python
# --- 1. DEFINIMOS LAS FUNCIONES DE DISTANCIA ---


def euclidean_distance(p1, p2):
    """Calcula la distancia Euclídea entre dos puntos (x, y)"""
    return math.sqrt((p1[0] - p2[0]) ** 2 + (p1[1] - p2[1]) ** 2)


def manhattan_distance(p1, p2):
    """Calcula la distancia Manhattan entre dos puntos (x, y)"""
    return abs(p1[0] - p2[0]) + abs(p1[1] - p2[1])


# --- 2. FUNCIÓN PRINCIPAL KNN MODIFICADA ---


def predict_color(points, new_point, k, dist_fn=euclidean_distance):
    distances = []

    for x, y, color in points:
        point_coords = (x, y)

        # Usamos la función de distancia que nos pasen por parámetro
        dist = dist_fn(point_coords, new_point)

        # Regla especial del enunciado: si la distancia es 0, ese color gana de inmediato
        if dist == 0:
            return color

        distances.append((dist, color))

    # Ordenar las distancias de menor a mayor y quedarnos con los k vecinos
    distances.sort(key=lambda item: item[0])
    k_nearest = distances[:k]

    # --- VOTO PONDERADO ---
    # Usaremos un diccionario tradicional para sumar los pesos de cada color
    color_weights = {}

    for dist, color in k_nearest:
        # El peso es inversamente proporcional a la distancia
        weight = 1 / dist

        # Acumulamos el peso para el color correspondiente
        if color in color_weights:
            color_weights[color] += weight
        else:
            color_weights[color] = weight

    # Encontrar el color con el peso acumulado más grande
    # max() busca en el diccionario el elemento con el valor (weight) más alto
    majority_color = max(color_weights, key=color_weights.get)

    return majority_color


# --- 3. PRUEBA DE EJECUCIÓN (Tu caso de prueba) ---
points = [
    (1.5, 1.8, "red"),
    (2.0, 2.1, "red"),
    (2.2, 1.9, "red"),
    (1.8, 2.3, "red"),
    (2.1, 2.4, "red"),
    (2.7, 1.8, "green"),
    (3.0, 2.2, "green"),
    (3.2, 1.9, "green"),
    (2.8, 2.4, "green"),
    (3.1, 2.5, "green"),
    (3.3, 2.8, "blue"),
    (3.5, 3.1, "blue"),
    (3.7, 2.9, "blue"),
    (3.4, 3.3, "blue"),
    (3.6, 3.2, "blue"),
]

# Nota: Corregí tu tupla (2.3, 1, 5) a (2.3, 1.5) porque estamos en un espacio 2D (X, Y)
nuevo_punto = (2.3, 1.5)
k = 6

# Prueba 1: Usando Distancia Euclídea (por defecto)
mali_euclidean = predict_color(points, new_point=nuevo_punto, k=k)
print(f"Predicción Euclídea: {mali_euclidean}")

# Prueba 2: Usando Distancia Manhattan (pasándola como parámetro)
mali_manhattan = predict_color(points, new_point=nuevo_punto, k=k, dist_fn=manhattan_distance)
print(f"Predicción Manhattan: {mali_manhattan}")
````

## Pig

Create a program that simulates the Pig dice game for multiple players. 

Pig Game Basics:
- Players take turns rolling a six-sided die.
- On their turn, a player can choose to continue rolling or hold.
- If they roll a 1, they score nothing, and their turn ends.
- If they roll 2-6, they can "hold" and add the sum of their rolls to their total score or risk rolling again.
- If a player holds, the turn goes to the next player.
- The first player to reach or exceed 100 points wins.

Implement the basic game for $N$ players.

For example, given players Alice, Bob and Charlie:

- Turn 1: Alice
    - Rolls: 4 → Turn total = 4
    - Decides to roll again
    - Rolls: 3 → Turn total = 7
    - Decides to hold → Alice total = 7

- Turn 2: Bob
    - Rolls: 5 → Turn total = 5
    - Rolls again
    - Rolls: 1 → Pig! Turn ends → Bob total = 0

- Turn 3: Charlie
    - Rolls: 2 → Turn total = 2
    - Rolls again
    - Rolls: 6 → Turn total = 8
    - Holds → Charlie total = 8

- Turn 4: Alice
    - Rolls: 3 → Turn total = 3
    - Rolls again
    - Rolls: 2 → Turn total = 5
    - Holds → Alice total = 12  (this is adding the previous Alice total of 7 points to this turn's total 5)

- ETC.

````python
import random


def play_pig_game(player_names):
    """Simula el juego de dados Pig con opción de salida voluntaria (q)."""
    scores = {name: 0 for name in player_names}
    winning_score = 100
    game_over = False

    print("=" * 40)
    print(" ¡BIENVENIDOS AL JUEGO DEL PIG! ")
    print("=" * 40)
    print(f"Objetivo: Llegar a {winning_score} puntos.")
    print("💡 Nota: Puedes escribir 'q' o 'exit' en cualquier momento para terminar el juego.\n")

    while not game_over:
        for player in player_names:
            print(f"\n--- TURNO DE: {player.upper()} ---")
            print(f"Puntuación acumulada actual: {scores[player]} puntos")

            turn_total = 0
            turn_active = True

            while turn_active:
                roll = random.randint(1, 6)
                print(f"🎲 -> ¡Has sacado un {roll}!")

                if roll == 1:
                    print(
                        f"💥 ¡PIG! Sacaste un 1. Pierdes los puntos de este turno."
                    )
                    turn_total = 0
                    turn_active = False
                else:
                    turn_total += roll
                    print(f"   Total provisional del turno: {turn_total}")

                    # Verificar si ya gana sumando el turno activo
                    if scores[player] + turn_total >= winning_score:
                        print(
                            f"\n✨ ¡Llegaste al objetivo de {winning_score}!"
                        )
                        turn_active = False
                        break

                    # Capturar la decisión del jugador
                    choice = (
                        input(
                            f"[{player}] ¿Volver a tirar (r), plantarte (h) o salir (q)?: "
                        )
                        .strip()
                        .lower()
                    )

                    # --- NUEVA CONDICIÓN DE SALIDA VOLUNTARIA ---
                    if choice in ["q", "exit"]:
                        print("\n" + "=" * 40)
                        print("🚪 Saliendo del juego de forma ordenada...")
                        print("=" * 40)
                        print("Marcador parcial al salir:")
                        for name, score in scores.items():
                            print(f" - {name}: {score} puntos")
                        print("\n¡Gracias por jugar!")
                        return  # Finaliza por completo la función de golpe y de forma limpia

                    # Si introduce 'h' o cualquier otra cosa que no sea continuar ('r'), se planta
                    if choice != "r":
                        print(f"-> Decides plantarte con {turn_total} puntos.")
                        turn_active = False

            scores[player] += turn_total
            print(
                f" Marcador final de {player}: {scores[player]} puntos.\n"
            )

            if scores[player] >= winning_score:
                print("=" * 50)
                print(f"🏆 ¡¡¡{player.upper()} HA GANADO LA PARTIDA!!! 🏆")
                print("=" * 50)
                print("\nMarcador Final:")
                for name, score in scores.items():
                    print(f" - {name}: {score} puntos")
                game_over = True
                break


# Ejecución limpia
lista_jugadores = ["Alice", "Bob", "Charlie"]
play_pig_game(lista_jugadores)
````

## Pig AI

After establishing the basic game (previous exercise), enhance it with an AI player that uses the Monte Carlo tree search method for decision-making.

````python
import random
import math
import copy

# ==============================================================================
# 1. CLASES PARA MONTE CARLO TREE SEARCH (MCTS)
# ==============================================================================

class PigState:
    """Representa el estado exacto del juego para las simulaciones de la IA."""
    def __init__(self, scores, current_player_idx, turn_total, winning_score=100):
        self.scores = copy.deepcopy(scores) # Diccionario {nombre: puntuación}
        self.current_player_idx = current_player_idx
        self.turn_total = turn_total
        self.winning_score = winning_score

    def get_current_player(self, player_names):
        return player_names[self.current_player_idx]

    def is_terminal(self):
        """El estado es terminal si alguien ha alcanzado la puntuación de victoria."""
        return any(score >= self.winning_score for score in self.scores.values())

    def get_winner(self):
        for player, score in self.scores.items():
            if score >= self.winning_score:
                return player
        return None

    def clone(self):
        return PigState(self.scores, self.current_player_idx, self.turn_total, self.winning_score)


class MCTSNode:
    """Nodo del árbol de búsqueda de Monte Carlo."""
    def __init__(self, state, player_names, action=None, parent=None):
        self.state = state
        self.player_names = player_names
        self.action = action  # 'r' (roll) o 'h' (hold)
        self.parent = parent
        self.children = []
        self.visits = 0
        self.wins = 0  # Victorias acumuladas para el jugador que tomó la acción

    def is_fully_expanded(self):
        # En Pig, las únicas acciones posibles en un turno son 'r' y 'h'
        valid_actions = ['r', 'h']
        expanded_actions = [child.action for child in self.children]
        return len(expanded_actions) == len(valid_actions)

    def uct_select_child(self, exploration_constant=1.41):
        """Selecciona un nodo usando el criterio UCB1 (Upper Confidence Bound)."""
        best_score = -float('inf')
        best_child = None

        for child in self.children:
            if child.visits == 0:
                return child
            # Explotación + Exploración
            exploitation = child.wins / child.visits
            exploration = exploration_constant * math.sqrt(math.log(self.visits) / child.visits)
            uct_value = exploitation + exploration

            if uct_value > best_score:
                best_score = uct_value
                best_child = child
        return best_child


def mcts_search(root_state, player_names, ai_name, iterations=1500):
    """Ejecuta el algoritmo MCTS durante N iteraciones y devuelve la mejor acción."""
    root_node = MCTSNode(root_state.clone(), player_names)

    for _ in range(iterations):
        node = root_node
        state = root_state.clone()

        # 1. SELECCIÓN: Bajar por el árbol usando UCT mientras esté expandido y no sea terminal
        while node.is_fully_expanded() and not state.is_terminal():
            node = node.uct_select_child()
            # Aplicar la acción seleccionada al estado simulado
            state = apply_action_simulation(state, node.action, player_names)

        # 2. EXPANSIÓN: Si no es terminal, añadir un nuevo nodo hijo legal
        if not state.is_terminal():
            expanded_actions = [child.action for child in node.children]
            for act in ['r', 'h']:
                if act not in expanded_actions:
                    next_state = apply_action_simulation(state.clone(), act, player_names)
                    new_child = MCTSNode(next_state, player_names, action=act, parent=node)
                    node.children.append(new_child)
                    node = new_child
                    state = next_state
                    break

        # 3. SIMULACIÓN (Rollout): Simular la partida de forma aleatoria hasta el final
        while not state.is_terminal():
            # Estrategia de simulación rápida: arriesgar con moderación
            action = 'r' if state.turn_total < 20 else 'h'
            state = apply_action_simulation(state, action, player_names)

        # 4. RETROPROPAGACIÓN (Backpropagation): Subir los resultados al árbol
        winner = state.get_winner()
        while node is not None:
            node.visits += 1
            # Si el ganador de la simulación coincide con el jugador de la IA, sumamos un triunfo
            if winner == ai_name:
                node.wins += 1
            node = node.parent

    # Retornar la acción del hijo con más visitas (la opción más robusta y segura)
    best_action_node = max(root_node.children, key=lambda child: child.visits)
    return best_action_node.action


def apply_action_simulation(state, action, player_names):
    """Aplica una acción a un estado simulado controlando las probabilidades del dado."""
    current_player = state.get_current_player(player_names)

    if action == 'h':
        # Guardar puntos acumulados y pasar turno
        state.scores[current_player] += state.turn_total
        state.turn_total = 0
        state.current_player_idx = (state.current_player_idx + 1) % len(player_names)
    elif action == 'r':
        # Simular promedio estocástico o una tirada real de dado
        roll = random.randint(1, 6)
        if roll == 1:
            state.turn_total = 0
            state.current_player_idx = (state.current_player_idx + 1) % len(player_names)
        else:
            state.turn_total += roll
            if state.scores[current_player] + state.turn_total >= state.winning_score:
                state.scores[current_player] += state.turn_total
                state.turn_total = 0
    return state


# ==============================================================================
# 2. FLUJO PRINCIPAL DEL JUEGO MODIFICADO
# ==============================================================================

def play_pig_game_ai(player_names, ai_name="IA_MonteCarlo"):
    """Simula el juego de dados Pig incluyendo a un agente Inteligencia Artificial."""
    all_players = player_names + [ai_name]
    scores = {name: 0 for name in all_players}
    winning_score = 100
    game_over = False

    print("=" * 55)
    print(" ¡BIENVENIDOS AL PIG ENHANCED CON IA (MONTE CARLO)! ")
    print("=" * 55)
    print(f"Objetivo: Llegar a {winning_score} puntos.")
    print("Escribe 'q' para salir de la partida en tu turno.\n")

    while not game_over:
        for idx, player in enumerate(all_players):
            print(f"\n--- TURNO DE: {player.upper()} ---")
            print(f"Puntuación acumulada actual: {scores[player]} puntos")

            turn_total = 0
            turn_active = True

            while turn_active:
                # Si el turno es de la IA, toma el control automáticamente
                if player == ai_name:
                    if turn_total == 0:
                        # Forzar primer lanzamiento obligatorio por reglas
                        choice = 'r'
                    else:
                        # Crear el estado virtual y lanzar la búsqueda en el árbol de decisión
                        current_state = PigState(scores, idx, turn_total, winning_score)
                        print("🤖 La IA está analizando miles de futuros posibles en el árbol MCTS...")
                        choice = mcts_search(current_state, all_players, ai_name, iterations=2000)

                    print(f"🤖 [IA] Acción decidida automáticamente: {'Tirar (r)' if choice == 'r' else 'Plantarse (h)'}")
                else:
                    # Turno del jugador humano
                    choice = input(f"[{player}] ¿Volver a tirar (r), plantarte (h) o salir (q)?: ").strip().lower()

                if choice in ["q", "exit"] and player != ai_name:
                    print("\n🚪 Saliendo de la partida de forma ordenada...")
                    return

                # Ejecución de la jugada física del dado
                if choice == 'r':
                    roll = random.randint(1, 6)
                    print(f"🎲 -> ¡Sacado un {roll}!")

                    if roll == 1:
                        print(f"💥 ¡PIG! Sacaste un 1. Pierdes todo lo acumulado este turno.")
                        turn_total = 0
                        turn_active = False
                    else:
                        turn_total += roll
                        print(f"   Total provisional de este turno: {turn_total}")

                        if scores[player] + turn_total >= winning_score:
                            print(f"\n✨ ¡Llegaste al objetivo de {winning_score}!")
                            turn_active = False
                            break
                else:
                    # Plantarse ('h')
                    print(f"-> Se planta sumando {turn_total} puntos.")
                    turn_active = False

            scores[player] += turn_total
            print(f" Marcador final de {player}: {scores[player]} puntos.\n")

            if scores[player] >= winning_score:
                print("=" * 50)
                print(f"🏆 ¡¡¡{player.upper()} HA GANADO LA PARTIDA!!! 🏆")
                print("=" * 50)
                print("\nMarcador Final de la partida:")
                for name, score in scores.items():
                    print(f" - {name}: {score} puntos")
                game_over = True
                break


# --- Ejecutar el juego contra la Inteligencia Artificial ---
# Define tus jugadores humanos
mis_jugadores = ["Levi", "Alice"]

# Arrancar partida
play_pig_game_ai(mis_jugadores)
````
