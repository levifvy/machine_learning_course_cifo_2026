## Més proper

Defineix una funció `closest(v, n)` que, donat un vector i un nombre $N$, retorni l'element més proper a $N$. En cas d'empat, retorna qualsevol d'aquests elements.

````python
def closest(v,n):
    idx = np.abs(v - n).argmin()
    return v[idx]
````

````python
# Els propers "tests" haurien de passar un cop definida la funció

assert closest([1, 2, 3, 4], 2) == 2
assert closest([1, 2, 3, 4], 5) == 4
assert closest([1, 2, 3, 4], -3) == 1
assert closest([1, 2, 2, 3, 4], 2) == 2
assert closest([1, 2, 2, 3, 4], 2.4) == 2
assert closest([1, 2, 2, 3, 4], 2.6) == 3
````

## Escalat Min-Max

Implementa una funció `minmax_scale(values)` que reescali una llista (o array de NumPy) de valors numèrics de manera que tots els valors estiguin dins del rang **[0, 1]**. Concretament, el nombre més petit tindrà el valor de $0$, mentre que el més gran tindrà el valor de $1$. La resta de nombres es trobaran entremig.

Aquest tipus de normalització és habitual en aprenentatge automàtic, ja que assegura que totes les característiques contribueixen de manera equitativa al procés d'entrenament d'un model, independentment de la seva escala original.

Donat un array de valors $X$, la fórmula d'escalat Min-Max per a cadascun dels seus valors $x_i$ és:

$x_i\_scaled = (x_i - min(X)) / (max(X) - min(X))$

## Extraterrestres

Els extraterrestres estan transmetent missatges binaris — arrays de 100 elements que contenen només 0s i 1s.

No obstant això, el seu missatge real està amagat dins del senyal: sempre comença i acaba amb 1, i tot el que hi ha abans del primer 1 i després de l'últim 1 és només soroll.

Per exemple, donat el missatge `[0, 0, 0, 1, 0, 1, 1, 0, 0, 1, 0, 0]`, el missatge real que has d'extreure és el rang des del primer 1 fins a l'últim, ambdós inclosos `[1, 0, 1, 1, 0, 0, 1]`.

````python
# Funció que genera un missatge aleatori d'extraterrestres

def get_aliens_message():
    # Es crea un missatge on la majoria són zeros
    message = np.zeros(100, dtype=int)

    # S'introdueix el missatge real
    start_index = np.random.randint(10, 40)
    end_index = np.random.randint(60, 90)

    message[start_index:end_index] = np.random.randint(2, size=end_index-start_index)
    message[start_index] = 1
    message[end_index] = 1
    return message
````

````python
message = get_aliens_message()
````

Si la matriu d'entrada té dimensions $m · n$, la matriu resultant tindrà dimensions $(m−1) · (n−1)$.

Per exemple, per a la matriu d'entrada

    [4, 7, 2, 5]

[

    [8 8 3]

    [8 8 5]
]

**Explicació**:
- La funció va lliscant una finestra 2×2 d'un pas en pas.
- Per a la regió superior esquerra de 2×2 que conté els valors 1, 3, 5 i 6, el valor màxim és 6. Això dona el primer valor a la part superior esquerra de la matriu retornada.
- Això es repeteix per tota la matriu per a cada regió 2×2.
